## Partie 9 : Gestion des valeurs manquantes 
### 1. Afficher le nombre de valeurs manquantes par colonne

```python
#afficher le nombre de valeurs manquantes par colonne
valeurs_manquantes = df.isnull().sum()
print("Nombre de valeurs manquantes par colonne:")
print(valeurs_manquantes)
```

### 2. Afficher le nombre total des valeurs manquantes

```python
#afficher le nombre total des valeurs manquantes
total_manquantes = df.isnull().sum().sum()
print(f"Nombre total de valeurs manquantes dans le dataframe: {total_manquantes}")
```

3. Afficher le taux de valeurs manquantes

```python
#afficher le taux de valeurs manquantes
taux_manquantes_global = (df.isnull().sum().sum() / (df.shape[0] * df.shape[1])) * 100
print(f"Taux global de valeurs manquantes: {taux_manquantes_global:.2f}%\n")
```

4. Afficher les lignes contenant des valeurs manquantes

```python
#afficher les lignes contenant des valeurs manquantes
lignes_avec_manquantes = df[df.isnull().any(axis=1)]
print("Lignes contenant des valeurs manquantes:")
print(lignes_avec_manquantes)
```

### 5. Pour la température et l’humidité, remplacer les valeurs manquantes par les moyennes

```python
#remplacer les valeurs manquantes par les moyennes pour temperature et humidite_relative

# Calculer les moyennes
moyenne_temperature = df['temperature'].mean()
moyenne_humidite = df['humidite_relative'].mean()

print(f"Moyenne - temperature: {moyenne_temperature:.2f}")
print(f"Moyenne - humidite_relative: {moyenne_humidite:.2f}\n")

# Remplacer les NaN par les moyennes
df['temperature'] = df['temperature'].fillna(moyenne_temperature)
df['humidite_relative'] = df['humidite_relative'].fillna(moyenne_humidite)


print("Après remplissage:")
print(f"Valeurs manquantes - temperature: {df['temperature'].isnull().sum()}")
print(f"Valeurs manquantes - humidite_relative: {df['humidite_relative'].isnull().sum()}")
```

### 6. Pour la consommation, remplacer les valeurs manquantes par la médiane

```python
# Calculer la médiane
mediane_consommation = df['consommation'].median()

print(f"Médiane - consommation: {mediane_consommation:.2f}\n")

#remplacer les valeurs manquantes de consommation par la médiane
df['consommation'] = df['consommation'].fillna(mediane_consommation)

print("Après remplissage:")
print(f"Valeurs manquantes - consommation: {df['consommation'].isnull().sum()}")
```

### 7. Pour "etat", remplacer les valeurs manquantes par "INCONNU"

```python
#remplacer les valeurs manquantes de etat par "INCONNU"
print("Avant remplissage:")
print(f"Valeurs manquantes - etat: {df['etat'].isnull().sum()}\n")

# Remplacer les NaN par "INCONNU"
df['etat'] = df['etat'].fillna("INCONNU")

print("Après remplissage:")
print(f"Valeurs manquantes - etat: {df['etat'].isnull().sum()}\n")

print("Valeurs uniques dans la colonne etat:")
print(df['etat'].value_counts())
```
