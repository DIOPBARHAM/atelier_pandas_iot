### 5. Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de mesure 12h, 13h, 14h et 15h

```python
import pandas as pd
#creer une serie avec des indices personnalisés
s1 = pd.Series(['18', '22', '25', '28'], index=['12h', '13h', '14h', '15h'])
print(f"serie avec indices personnalisés:\n{s1}\n")
#creer une serie à partir d'un dictionnaire
d = {'12h': 18, '13h': 22, '14h': 25, '15h': 28}
s2 = pd.Series(d)
print(f"serie à partir du dictionnaire:\n{s2}\n")
```

### 6. Afficher la 1ère et la dernière valeur

```python
import pandas as pd
s1 = pd.Series(['18', '22', '25', '28'], index=['12h', '13h', '14h', '15h'])
d = {'12h': 18, '13h': 22, '14h': 25, '15h': 28}
s2 = pd.Series(d)
# accès par position numérique, peu importe le label
print(f"Première valeur de la série avec indice personnalisé: {s1.iloc[0]}")

print(f"Dernière valeur de la série avec indice personnalisé: {s1.iloc[len(s1)-1]}")    

print(f"Première valeur de la série avec dictionnaire: {s2.iloc[0]}")

print(f"Dernière valeur de la série avec dictionnaire: {s2.iloc[len(s2)-1]}")


```
